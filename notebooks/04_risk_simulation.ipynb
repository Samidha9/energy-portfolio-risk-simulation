```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

results = pd.read_csv(
    "../data/processed/forecast_results.csv",
    parse_dates=["date"]
)

errors = results["error"].values
```

```python
plt.hist(errors, bins=30)
plt.xlabel("Forecast Error ($/MWh)")
plt.ylabel("Frequency")
plt.title("Power Price Forecast Error Distribution")
plt.show()
```
